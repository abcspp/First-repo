### Tokenization

```python
# Step 1: Import Libraries
import nltk
import string

# Step 2: Download Required Packages
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('stopwords')
nltk.download('omw-1.4')

# Step 3: Import Modules
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

from nltk.stem import WordNetLemmatizer
from nltk.corpus.reader.wordnet import NOUN, VERB, ADJ, ADV

from nltk.stem import PorterStemmer
from nltk.stem.snowball import SnowballStemmer
from nltk.stem import LancasterStemmer
```

```python
text = "Real madrid is set to win the UCL for the season. Benzema might win Balon dor. Salah might be the runner running up went children faster. "
```

```python
# Step 4: Sentence Tokenization
tokens_sents = sent_tokenize(text)
print("Sentence Tokens:")
print(tokens_sents)
```

```python
# Step 5: Word Tokenization
tokens_words = word_tokenize(text)
print("\nWord Tokens:")
print(tokens_words)
```

```python
# Step 6: Stopword & Punctuation Removal
sw_nltk = stopwords.words('english')

filtered_words = [word for word in tokens_words
                  if word.lower() not in sw_nltk
                  and word not in string.punctuation]

print("\nAfter Removing Stopwords & Punctuation:")
print(filtered_words)
```

```python
# Step 7: POS Tagging
pos_tags = nltk.pos_tag(filtered_words)

print("\nPOS Tags:")
print(pos_tags)
```

```python
# Step 7: Function to Convert POS Tags
def get_wordnet_pos(tag):
    
    if tag.startswith('J'):
        return ADJ
    
    elif tag.startswith('V'):
        return VERB
    
    elif tag.startswith('N'):
        return NOUN
    
    elif tag.startswith('R'):
        return ADV
    
    else:
        return NOUN
```

### Lemmatization


```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = []

for word, tag in pos_tags:
    
    wordnet_pos = get_wordnet_pos(tag)
    
    lemma_word = lemmatizer.lemmatize(word, pos=wordnet_pos)
    
    lemmatized_words.append(lemma_word)

print("\nLemmatized Words:")
print(lemmatized_words)
```

```python
# Porter Stemmer
ps = PorterStemmer()

porter_stemmed = [ps.stem(word)
                  for word in filtered_words]

print("\nPorter Stemmer Output:")
print(porter_stemmed)
```

```python
# Snowball Stemmer
ss = SnowballStemmer("english")

snowball_stemmed = [ss.stem(word)
                    for word in filtered_words]

print("\nSnowball Stemmer Output:")
print(snowball_stemmed)
```

```python
# Lancaster Stemmer
ls = LancasterStemmer()

lancaster_stemmed = [ls.stem(word)
                     for word in filtered_words]

print("\nLancaster Stemmer Output:")
print(lancaster_stemmed)
```

```python
# Step 11: Join Words into Final Sentence
final_sentence = " ".join(porter_stemmed)

print("\nFinal Sentence:")
print(final_sentence)
```
