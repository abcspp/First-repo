```python
import nltk
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
```

```python
text = """
Natural Language Processing is a field of Artificial Intelligence.
It helps computers understand human language.
Text analytics is very useful in machine learning.
"""
```

```python
print("Original Text:\n")
print(text)
```

```python
# 1. Tokenization
tokens = word_tokenize(text)
print("Tokenized Words",tokens)
```

```python
# 2. POS Tagging
pos = pos_tag(tokens)
print("POS Tagging")
print(pos)
```

```python
# 3. Stopwords Removal
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words
]
print("After Stopwords Removal")
print(filtered_words)
```

```python
# 4. Stemming
ps = PorterStemmer()
stemmed_words = [
    ps.stem(word) for word in filtered_words
]
print("Stemmed Words")

print(stemmed_words)
```

```python
# 5. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]
print("Lemmatized Words")

print(lemmatized_words)
```

```python
# 6. Term Frequency (TF)
documents = [text]
cv = CountVectorizer()
tf = cv.fit_transform(documents)
tf_df = pd.DataFrame(
    tf.toarray(),
    columns=cv.get_feature_names_out()
)
print("Term Frequency (TF)")
print(tf_df)
```

```python
# 7. TF-IDF
tfidf = TfidfVectorizer()
tfidf_result = tfidf.fit_transform(documents)
tfidf_df = pd.DataFrame(
    tfidf_result.toarray(),
    columns=tfidf.get_feature_names_out()
)
print("TF-IDF")
print(tfidf_df)
```
